# Q35...Find the average Billing_Amount for each Diagnosis.

```python
df.groupby('Diagnosis')['Billing Amount'].mean().round(2)
```

```python
x=df.groupby('Diagnosis')['Billing Amount'].mean().index
y=df.groupby('Diagnosis')['Billing Amount'].mean().values

def show_number(p):
    return round(p * sum(y) / 100, 2 )
plt.title('Average Billing Amount Per Diagnosis')
plt.pie(y, labels=x,autopct=show_number)
plt.show()
```

# Q36...Find the average Length_Of_Stay for each Diagnosis.

```python
df.groupby('Diagnosis')['Length_Of_Stay'].mean()
```

```python
y=df.groupby('Diagnosis')['Length_Of_Stay'].mean().dt.days.values  # for changing days value in int data type
x=df.groupby('Diagnosis')['Length_Of_Stay'].mean().index

plt.xticks(rotation=45)
plt.bar(x,y,color='r')
plt.show()
```

# Q37...Find the average Feedback for each Doctor.

```python
df.groupby('Doctor')['Feedback'].mean()
```

```python
x=df.groupby('Doctor')['Feedback'].mean().index
y=df.groupby('Doctor')['Feedback'].mean().values

def value(p):
    return round(p* sum(y)/100,3)
plt.title(' feedback for doctor')
plt.pie(y,labels=x,autopct=value)
plt.show()
```

# Q38...Find the total billing amount generated by each Doctor.

```python
df.groupby('Doctor')['Billing Amount'].sum()
```

```python
import seaborn as sns
```

```python
x=df.groupby('Doctor')['Billing Amount'].sum().index
y=df.groupby('Doctor')['Billing Amount'].sum().values

for i in range(len(x)):
    plt.text(i, y[i], int(y[i]), ha='center')

plt.title('Billing Amount per Doctor')
plt.ylabel('Amount in 10 Million')
plt.xticks(rotation=45)
sns.barplot(x=x,y=y)
plt.show()
```
